## Load a CSV Dataset

This task demonstrates how to load a CSV dataset using Pandas and display its first and last records.

```python
import pandas as pd
```

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data"

df = pd.read_csv(
    url,
    header=None,
    names=[
        "SepalLengthCm",
        "SepalWidthCm",
        "PetalLengthCm",
        "PetalWidthCm",
        "Species"
    ]
)

print("CSV dataset loaded successfully!")
```
